```python
import graphical_sampling as gs
import numpy as np
import matplotlib.pyplot as plt
```

```python
N1, N2 = 8, 8
N = N1*N2
n = 4
rng = gs.random.rng()
# coords = rng.rand_coord((N, 2))
coords = rng.grid_coords((N1, N2))
inclusions = rng.equal_inclusions(n, N)
pop = gs.Population(coords, inclusions, n=n)
```

```python
plt.scatter(*pop.coords.T)
for i, coord in enumerate(pop.coords):
    plt.text(*(coord+0.01).T, s=f'{i}')
```

```python
fbn = gs.clustering.IPBalancedOTNMeans(n)
fbn.fit(pop)
fbn.plot(mode='hard')
```
