```python
import os
import glob
import numpy as np
import graphical_sampling as gs
import pandas as pd
```

```python
N = 1000
n = 4
num_zones = 4
rng = gs.random.rng()
coords = rng.rand_coord((N, 2))
variable = np.random.rand(N) * 10
# inclusions = rng.unequal_inclusions(n, N)
inclusions = variable + np.random.rand(N)
pop = gs.Population(coords, inclusions, n=n, variable=variable)
```

```python
fbn = gs.clustering.IPBalancedOTNMeans(n)
fbn.fit(pop)
fbn.fit_zones(num_zones)
```

```python
fbn.plot(mode='soft')
```

```python
original_cluster = fbn.clusters
```

```python
order = gs.Order.from_clusters(pop, original_cluster, strategy='spiral')
```

```python
fbn.clusters = order.clusters
```

```python
fbn.plot(mode='soft')
```
