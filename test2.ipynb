```python
import numpy as np

import graphical_sampling as gs
```

```python
N = 1000
n = 6
rng = gs.random.rng()
coords = rng.rand_coord((N, 2))
inclusions = rng.unequal_inclusions(n, N)
pop = gs.Population(coords, inclusions, n=n)
```

```python
design = gs.Design(pop, num_zones=n)
design.show(save=True)
```

```python
fbn = gs.clustering.FIPBalancedNMeans(n=n)
fbn.fit(pop)
```

```python
fbn.fit_zones(num_zones=4)
```

```python
fbn.plot(mode='soft', show_zone_sub_hulls=True, show_zone_labels=True)
```

```python
order = gs.Order.from_clusters(pop, fbn.clusters)
```

```python
order.get()
```

```python
design2 = gs.Design(pop, num_zones=4, order=order)
```

```python
design2.show(save=True)
```
