```python
import pandas as pd

# Load your data
df = pd.read_csv('simulation_results/n-Means_on_5_popu.csv')

# 1. Replace the `prob` column (remove it, embed its value in `coord`)
# We combine `coord` and `prob` to form new `coord`
def update_coord(row):
    if 'eq' in row['coord']:
        return row['coord'].replace('eq', '') + '_eq'
    elif 'uneq' in row['coord']:
        return row['coord'].replace('uneq', '') + '_uneq'
    else:
        # fallback if pattern not found
        return row['coord']

df['coord'] = df.apply(update_coord, axis=1)

# 2. Remove old `prob` column
df = df.drop(columns=['prob'])

# 3. Sort by coord, n, zones, exp_IB
df_sorted = df.sort_values(
    by=['coord', 'n', 'zones', 'exp_LB']
)

# 4. Optional: Reset index
df_sorted = df_sorted.reset_index(drop=True)

# 5. Save back to CSV (overwrite or give new name)
df_sorted.to_csv('simulation_results/Desings_on_5_popu_sorted.csv', index=False)
```

```python
df_sorted.rename(
    columns={
        'coord': 'pop',
        'prob': 'p',
        'n': 'n',
        'zones': 'zns',
        'zone_mode': 'size',
        'sort_method': 'sort',
        'zonal_sort': 'zsort',
        'exp_density': 'edens',
        'var_density': 'vdens',
        'min_density': 'mdens',
        'median_density': 'meddens',
        'max_density': 'maxdens',
        'exp_IB': 'eIB',
        'var_IB': 'vIB',
        'min_IB': 'mIB',
        'median_IB': 'medIB',
        'max_IB': 'maxIB',
        'exp_LB': 'eLB',
        'var_LB': 'vLB',
        'min_LB': 'mLB',
        'median_LB': 'medLB',
        'max_LB': 'maxLB',
    },
    inplace=True
)
```
