# Go to the folder where the terminal script is saved
cd /home/bardia/projects/graphical-sampling/simulations_abc/terminal_running

# Run the script normally in terminal
python run_abc_mu284_terminal.py

# Start a protected tmux session named abc_mu284
tmux new -s abc_mu284

# After entering tmux, run the script
python run_abc_mu284_terminal.py

# Leave tmux but keep the script running
# Press Ctrl+B, then press D

# Come back to the running tmux session
tmux attach -t abc_mu284

# Stop the running script
# Press Ctrl+C

# Kill the whole tmux session
tmux kill-session -t abc_mu284

# Run the script in background without tmux and save output to log
nohup python -u run_abc_mu284_terminal.py > abc_mu284.log 2>&1 &

# See live printed results from the log file
tail -f abc_mu284.log
